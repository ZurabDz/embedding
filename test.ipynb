```python
import os
from google import genai
from google.genai import types

# Key comes from the environment (see .dotenv) — never hardcode it here.
client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])
```

```python
response = client.models.embed_content(
    model="gemini-embedding-001",
    contents=["გარეთ წვიმდა ძალიან გუშინ", "დღეს მზიანი და თბილი ამინდია"],
    config=types.EmbedContentConfig(task_type="RETRIEVAL_QUERY"),
)

print(len(response.embeddings), "embeddings, dim", len(response.embeddings[0].values))
```

```python
embedding = response.embeddings[0].values
```

```python
embedding2 = response.embeddings[1].values
```

```python
from scipy.spatial.distance import cosine
```

```python
cosine(embedding, embedding2)
```
